```r
install.packages("emmeans")
library(emmeans)
##################### QUESTÃO 1 #####################
### LETRA A ###
dose <- factor(rep(c(20,30,40,50), each = 6))
observações <- c(24,28,37,30,33,25,
                37,44,31,35,38,35,
                42,47,52,38,37,36,
                45,41,47,40,49,47)
dados <- data.frame(dose, observações)

plot(observações, dose,pch = 19, xlab = "Observações", ylab = "Dose em gramas")

### LETRA B ###
mean(observações)
mean20 = mean(c(24,28,37,30,33,25))
mean30 = mean(c(37,44,31,35,38,35))
mean40 = mean(c(42,47,52,38,37,36))
mean50 = mean(c(45,41,47,40,49,47))

### LETRA C ###
#Usando a tabela anova, que fornece as somas dos quadrados que precisamos:
anova <- aov(observações ~ dose)
summary(anova)
#Sum Sq = SQTrat ; Sum Sq Residuals = SQErro ; SQTot = Soma dos dois
#F value = F0 e Fcrit = qf(1-0,01;4-1;24-4)
qf(0.99,3,20)
#A estimativa do desvio padrão das observações é a raiz quadrada da divisão entre
#SQErro/GLerro
sigma <- summary(anova)[[1]]["Residuals", "Mean Sq"]
sqrt(sigma)

### LETRA D ###
library(emmeans)
emmeans(anova, ~ dose, level = 0.90)
```
